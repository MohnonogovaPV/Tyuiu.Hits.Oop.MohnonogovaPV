<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();

    public int NumberOfDoors { get; set; }
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
// Делегаты
public delegate void TransportAction();
public delegate void TransportEventHandler(string message);

// Базовый класс с событиями
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    
    // Событие
    public event TransportEventHandler TransportEvent;
    
    protected virtual void OnTransportEvent(string message)
    {
        TransportEvent?.Invoke(message);
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"{Manufacturer} {Model}");
    }
    
    public void Start()
    {
        OnTransportEvent($"{Manufacturer} {Model} запущен");
    }
}

public class Car : Transport
{
    public int Doors { get; set; }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"{Manufacturer} {Model} - {Doors} дверей");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }
    
    public override void DisplayInfo()
    {
        string carrier = HasCarrier ? "с багажником" : "без багажника";
        Console.WriteLine($"{Manufacturer} {Model} - {carrier}");
    }
}

// Использование
Console.WriteLine("=== Делегаты ===");
Transport car = new Car { Model = "Camry", Manufacturer = "Toyota", Doors = 4 };
Transport bike = new Bike { Model = "Mountain", Manufacturer = "Giant", HasCarrier = true };

// Делегаты
TransportAction carAction = car.DisplayInfo;
TransportAction bikeAction = bike.DisplayInfo;

carAction();
bikeAction();

Console.WriteLine("\n=== Лямбда-выражения ===");
List<Transport> transports = new List<Transport>
{
    new Car { Model = "Camry", Manufacturer = "Toyota", Doors = 4 },
    new Bike { Model = "Mountain", Manufacturer = "Giant", HasCarrier = true },
    new Car { Model = "Civic", Manufacturer = "Honda", Doors = 4 },
    new Bike { Model = "Street", Manufacturer = "Harley", HasCarrier = false }
};

// Лямбда для фильтрации
var toyotaCars = transports.Where(t => t.Manufacturer == "Toyota");
Console.WriteLine("Toyota автомобили:");
foreach (var transport in toyotaCars)
{
    transport.DisplayInfo();
}

// Лямбда для сортировки
var sortedTransports = transports.OrderBy(t => t.Manufacturer);
Console.WriteLine("\nСортировка по производителю:");
foreach (var transport in sortedTransports)
{
    transport.DisplayInfo();
}

Console.WriteLine("\n=== События ===");
// Подписка на события
car.TransportEvent += (message) => Console.WriteLine($"Событие автомобиля: {message}");
bike.TransportEvent += (message) => Console.WriteLine($"Событие мотоцикла: {message}");

// Вызов событий
car.Start();
bike.Start();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
// Делегаты для животных
public delegate void AnimalAction();
public delegate void AnimalEventHandler(string message);

// Базовый класс Animal с событиями
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    
    // Событие
    public event AnimalEventHandler AnimalEvent;
    
    protected virtual void OnAnimalEvent(string message)
    {
        AnimalEvent?.Invoke(message);
    }
    
    public virtual void Speak()
    {
        Console.WriteLine($"{Name} издает звук");
    }
    
    public void Feed()
    {
        OnAnimalEvent($"{Name} покормлен");
    }
}

public class Dog : Animal
{
    public override void Speak()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
}

public class Cat : Animal
{
    public override void Speak()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
}

public class Bird : Animal
{
    public override void Speak()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
}

// Использование
Console.WriteLine("=== Делегаты для животных ===");
Animal dog = new Dog { Name = "Бобик", Age = 3 };
Animal cat = new Cat { Name = "Мурка", Age = 2 };
Animal bird = new Bird { Name = "Кеша", Age = 1 };

// Делегаты
AnimalAction dogAction = dog.Speak;
AnimalAction catAction = cat.Speak;
AnimalAction birdAction = bird.Speak;

dogAction();
catAction();
birdAction();

Console.WriteLine("\n=== Лямбда-выражения для животных ===");
List<Animal> animals = new List<Animal>
{
    new Dog { Name = "Бобик", Age = 3 },
    new Cat { Name = "Мурка", Age = 2 },
    new Bird { Name = "Кеша", Age = 1 },
    new Dog { Name = "Шарик", Age = 4 },
    new Cat { Name = "Васька", Age = 3 }
};

// Лямбда для фильтрации
var olderAnimals = animals.Where(a => a.Age > 2);
Console.WriteLine("Животные старше 2 лет:");
foreach (var animal in olderAnimals)
{
    animal.Speak();
}

// Лямбда для группировки
var groupedAnimals = animals.GroupBy(a => a.GetType().Name);
Console.WriteLine("\nГруппировка по типу:");
foreach (var group in groupedAnimals)
{
    Console.WriteLine($"{group.Key}: {group.Count()} животных");
    foreach (var animal in group)
    {
        Console.WriteLine($"  - {animal.Name}");
    }
}

Console.WriteLine("\n=== События для животных ===");
// Подписка на события с лямбдами
dog.AnimalEvent += (message) => Console.WriteLine($"Уведомление: {message}");
cat.AnimalEvent += (message) => Console.WriteLine($"Уведомление: {message}");
bird.AnimalEvent += (message) => Console.WriteLine($"Уведомление: {message}");

// Вызов событий
dog.Feed();
cat.Feed();
bird.Feed();

// Дополнительные лямбда-операции
Console.WriteLine("\n=== Дополнительные операции ===");
animals.ForEach(animal => {
    if (animal.Age < 3)
    {
        Console.WriteLine($"{animal.Name} - молодой");
    }
});

var animalNames = animals.Select(a => a.Name).ToList();
Console.WriteLine($"\nИмена животных: {string.Join(", ", animalNames)}");
````
