Создать базовый класс PaymentMethod в C#, который будет представлять 
различные способы оплаты. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу PaymentMethod: 
• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты 
(MethodName), Минимальная сумма (MinAmount). 
• Методы: 
o ProcessPayment(decimal amount): метод для обработки платежа 
указанной суммы. 
o CheckMinimumAmount(decimal amount): метод для проверки 
минимальной суммы платежа. 
o GetPaymentDetails(): метод для получения деталей способа оплаты.

атрибуты,

Метод CheckMinimumAmount() должен быть переопределен для проверки

3. Наличные (CashPayment) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Место выдачи наличных

```
using System;

// Интерфейсы для множественного наследования
public interface ISecurity
{
    bool VerifySecurity();
}

public interface INotification
{
    void SendNotification(string message);
}

// Базовый класс PaymentMethod
public class PaymentMethod
{
    public int PaymentMethodId { get; set; }
    public string MethodName { get; set; }
    public decimal MinAmount { get; set; }
    public string Currency { get; set; }
    public bool IsActive { get; set; }

    public PaymentMethod(int id, string name, decimal minAmount)
    {
        PaymentMethodId = id;
        MethodName = name;
        MinAmount = minAmount;
        Currency = "RUB";
        IsActive = true;
    }

    public virtual void ProcessPayment(decimal amount)
    {
        if (CheckMinimumAmount(amount))
        {
            Console.WriteLine($"Обработка платежа {amount} {Currency} через {MethodName}");
        }
        else
        {
            Console.WriteLine($"Ошибка: сумма меньше минимальной {MinAmount} {Currency}");
        }
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"{MethodName} (мин. {MinAmount} {Currency})");
    }

    // Новый метод
    public void ToggleActive()
    {
        IsActive = !IsActive;
        Console.WriteLine($"{MethodName} теперь {(IsActive ? "активен" : "неактивен")}");
    }
}

// Простое наследование
public class OnlinePayment : PaymentMethod, ISecurity, INotification
{
    public string PaymentUrl { get; set; }
    public string GatewayName { get; set; }
    public bool SupportsRecurring { get; set; }

    public OnlinePayment(int id, string name, decimal minAmount, string paymentUrl, string gatewayName)
        : base(id, name, minAmount)
    {
        PaymentUrl = paymentUrl;
        GatewayName = gatewayName;
        SupportsRecurring = false;
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!IsActive) return;

        if (CheckMinimumAmount(amount) && VerifySecurity())
        {
            Console.WriteLine($"Онлайн платеж {amount} {Currency} через {GatewayName}");
            Console.WriteLine($"URL: {PaymentUrl}");
            SendNotification($"Платеж {amount} {Currency} выполнен");
        }
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Шлюз: {GatewayName}, URL: {PaymentUrl}");
    }

    // Новый метод
    public void EnableRecurring()
    {
        SupportsRecurring = true;
        Console.WriteLine("Рекуррентные платежи активированы");
    }

    // Реализация интерфейсов
    public bool VerifySecurity()
    {
        Console.WriteLine("Проверка SSL сертификата... OK");
        return true;
    }

    public void SendNotification(string message)
    {
        Console.WriteLine($"Email: {message}");
    }
}

// Сложное наследование
public class BankTransfer : PaymentMethod, INotification
{
    public string BankData { get; set; }
    public decimal CommissionRate { get; set; }
    public string BankCode { get; set; }
    public string AccountNumber { get; set; }

    public BankTransfer(int id, string name, decimal minAmount, string bankData, 
                       decimal commissionRate, string bankCode, string accountNumber)
        : base(id, name, minAmount)
    {
        BankData = bankData;
        CommissionRate = commissionRate;
        BankCode = bankCode;
        AccountNumber = accountNumber;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        return (amount + amount * CommissionRate) >= MinAmount;
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!IsActive) return;

        decimal commission = amount * CommissionRate;
        decimal total = amount + commission;

        if (CheckMinimumAmount(amount))
        {
            Console.WriteLine($"Банковский перевод {amount} {Currency}");
            Console.WriteLine($"Комиссия: {commission} {Currency}, Итого: {total} {Currency}");
            Console.WriteLine($"Счет: {AccountNumber}");
            SendNotification($"Перевод {amount} {Currency} выполнен");
        }
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Банк: {BankData}, Комиссия: {CommissionRate * 100}%");
    }

    // Новый метод
    public void ValidateAccount()
    {
        if (AccountNumber.Length >= 10)
            Console.WriteLine("Счет валиден");
        else
            Console.WriteLine("Ошибка: неверный номер счета");
    }

    // Реализация интерфейса
    public void SendNotification(string message)
    {
        Console.WriteLine($"SMS: {message}");
    }
}

// Множественное наследование
public class CashPayment : PaymentMethod, ISecurity
{
    public string CashPickupPoint { get; set; }
    public string OperatorName { get; set; }
    public string LocationAddress { get; set; }

    public CashPayment(int id, string name, decimal minAmount, string cashPickupPoint, string operatorName)
        : base(id, name, minAmount)
    {
        CashPickupPoint = cashPickupPoint;
        OperatorName = operatorName;
        LocationAddress = "Основное отделение";
    }

    public override void ProcessPayment(decimal amount)
    {
        if (!IsActive) return;

        if (CheckMinimumAmount(amount) && VerifySecurity())
        {
            Console.WriteLine($"Оплата наличными {amount} {Currency}");
            Console.WriteLine($"Место: {CashPickupPoint}, Оператор: {OperatorName}");
        }
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Пункт: {CashPickupPoint}, Адрес: {LocationAddress}");
    }

    // Новый метод
    public void PrintReceipt(decimal amount)
    {
        Console.WriteLine($"Чек: {amount} {Currency} - {CashPickupPoint}");
    }

    // Реализация интерфейса
    public bool VerifySecurity()
    {
        Console.WriteLine("Проверка кассового аппарата... OK");
        return true;
    }
}

// Демонстрация
public class Program
{
    public static void Main()
    {
        
    }
}
Console.WriteLine("=== ДЕМОНСТРАЦИЯ СИСТЕМЫ ОПЛАТЫ ===\n");

        // Простое наследование
        Console.WriteLine("1. ONLINE PAYMENT:");
        OnlinePayment online = new OnlinePayment(1, "Интернет-оплата", 10, 
            "https://pay.site.com", "Stripe");
        online.GetPaymentDetails();
        online.ProcessPayment(100);
        online.EnableRecurring();
        Console.WriteLine();

        // Сложное наследование
        Console.WriteLine("2. BANK TRANSFER:");
        BankTransfer bank = new BankTransfer(2, "Банковский перевод", 50, 
            "Сбербанк", 0.015m, "044525225", "40702810000000012345");
        bank.GetPaymentDetails();
        bank.ValidateAccount();
        bank.ProcessPayment(200);
        Console.WriteLine();

        // Множественное наследование
        Console.WriteLine("3. CASH PAYMENT:");
        CashPayment cash = new CashPayment(3, "Наличные", 1, 
            "Касса магазина", "ИП Петров");
        cash.GetPaymentDetails();
        cash.ProcessPayment(50);
        cash.PrintReceipt(50);
        Console.WriteLine();

        // Демонстрация полиморфизма
        Console.WriteLine("4. ПОЛИМОРФИЗМ:");
        PaymentMethod[] payments = { online, bank, cash };
        
        foreach (var payment in payments)
        {
            Console.WriteLine($"--- {payment.MethodName} ---");
            payment.ProcessPayment(75);
            payment.ToggleActive();
        }

        Console.WriteLine("\n=== ПРОГРАММА ЗАВЕРШЕНА ===");
```
